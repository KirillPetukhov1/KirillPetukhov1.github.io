## Часть 1. Исследование дополнительных моделей

Ниже исследуются следующие алгоритмы:
- **Gradient Boosting (GBM)** - бустинг на деревьях решений
- **XGBoost** - оптимизированный градиентный бустинг
- **Ridge Regression** - линейная регрессия с L2-регуляризацией

Цель: найти модель с ошибкой **меньше**, чем у базового Random Forest.

````python
# Устанавливаем xgboost если не установлен
!pip install xgboost -q
````

````python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.linear_model import Ridge
from xgboost import XGBRegressor

# --- Ridge Regression ---
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(training_points, training_values)
preds_ridge = ridge_model.predict(test_points)
mae_ridge  = mean_absolute_error(test_values, preds_ridge)
rmse_ridge = np.sqrt(mean_squared_error(test_values, preds_ridge))
print(f"Ridge — MAE: {mae_ridge:,.0f}, RMSE: {rmse_ridge:,.0f}")

# --- Gradient Boosting ---
gbm_model = GradientBoostingRegressor(n_estimators=200, max_depth=5,
                                       learning_rate=0.1, random_state=111)
gbm_model.fit(training_points, training_values)
preds_gbm = gbm_model.predict(test_points)
mae_gbm  = mean_absolute_error(test_values, preds_gbm)
rmse_gbm = np.sqrt(mean_squared_error(test_values, preds_gbm))
print(f"GBM   — MAE: {mae_gbm:,.0f}, RMSE: {rmse_gbm:,.0f}")

# --- XGBoost ---
xgb_model = XGBRegressor(n_estimators=500, max_depth=6,
                          learning_rate=0.05, subsample=0.8,
                          colsample_bytree=0.8, random_state=111,
                          verbosity=0)
xgb_model.fit(training_points, training_values)
preds_xgb = xgb_model.predict(test_points)
mae_xgb  = mean_absolute_error(test_values, preds_xgb)
rmse_xgb = np.sqrt(mean_squared_error(test_values, preds_xgb))
print(f"XGB   — MAE: {mae_xgb:,.0f}, RMSE: {rmse_xgb:,.0f}")
````

````python
# Сводная таблица сравнения всех моделей
results = {
    'Линейная регрессия': {
        'MAE': mean_absolute_error_linear_model,
        'RMSE': np.sqrt(mean_squared_error_linear_model)
    },
    'Ridge Regression': {'MAE': mae_ridge,   'RMSE': rmse_ridge},
    'Random Forest':    {'MAE': mean_absolute_error_random_forest_model,
                         'RMSE': np.sqrt(mean_squared_error_random_forest_model)},
    'Gradient Boosting':{'MAE': mae_gbm,     'RMSE': rmse_gbm},
    'XGBoost':          {'MAE': mae_xgb,     'RMSE': rmse_xgb},
}
results_df = pd.DataFrame(results).T.sort_values('RMSE')
results_df['MAE']  = results_df['MAE'].map('{:,.0f}'.format)
results_df['RMSE'] = results_df['RMSE'].map('{:,.0f}'.format)
print(results_df.to_string())
````

````python
# Визуализация предсказаний XGBoost (лучшая модель)
plt.figure(figsize=(7, 7))
plt.scatter(test_values, preds_xgb, s=1, c='green')
plt.plot([0, 6e6], [0, 6e6], 'r--')
plt.xlabel('Настоящая цена', fontsize=16)
plt.ylabel('Предсказанная цена', fontsize=16)
plt.title('XGBoost — предсказания vs реальные', fontsize=14)
plt.tight_layout()
plt.show()
````

## Часть 2. Оптимизация Random Forest (подбор гиперпараметров)

Используем `RandomizedSearchCV` для поиска лучших параметров модели Random Forest.
Цель: снизить RMSE ниже базового значения.

````python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestRegressor

# Пространство поиска гиперпараметров
param_dist = {
    'n_estimators':      [100, 200, 300, 500],
    'max_depth':         [None, 10, 20, 30, 40],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf':  [1, 2, 4],
    'max_features':      ['sqrt', 'log2', 0.5, 0.7],
    'bootstrap':         [True, False],
}

rf_base = RandomForestRegressor(random_state=111, n_jobs=-1)

# RandomizedSearchCV: 20 случайных комбинаций, 3-fold CV
random_search = RandomizedSearchCV(
    rf_base,
    param_distributions=param_dist,
    n_iter=20,
    cv=3,
    scoring='neg_root_mean_squared_error',
    random_state=42,
    n_jobs=-1,
    verbose=1
)
random_search.fit(training_points, training_values)

print("Лучшие параметры:", random_search.best_params_)
print("Лучший CV RMSE:", -random_search.best_score_)
````

````python
# Оцениваем лучшую модель на тестовой выборке
best_rf = random_search.best_estimator_
preds_best_rf = best_rf.predict(test_points)

mae_best_rf  = mean_absolute_error(test_values, preds_best_rf)
rmse_best_rf = np.sqrt(mean_squared_error(test_values, preds_best_rf))

print(f"Оптимизированный Random Forest:")
print(f"  MAE : {mae_best_rf:,.0f}")
print(f"  RMSE: {rmse_best_rf:,.0f}")
print()
print(f"Базовый Random Forest:")
print(f"  MAE : {mean_absolute_error_random_forest_model:,.0f}")
print(f"  RMSE: {np.sqrt(mean_squared_error_random_forest_model):,.0f}")
````

````python
# Важность признаков оптимизированной модели
feature_importances = pd.Series(
    best_rf.feature_importances_,
    index=training_points.columns
).sort_values(ascending=True)

plt.figure(figsize=(10, 6))
feature_importances.plot(kind='barh', color='steelblue')
plt.title('Важность признаков — Оптимизированный Random Forest', fontsize=14)
plt.xlabel('Важность', fontsize=12)
plt.tight_layout()
plt.show()
````

### Пошаговый алгоритм развёртывания ML-модели в виде REST API

#### Шаг 1. Сохранение обученной модели

#### Шаг 2. Создание веб-приложения (FastAPI)

```python
# app.py
from fastapi import FastAPI
from pydantic import BaseModel
import joblib
import numpy as np

app = FastAPI(title="House Price Predictor")
model = joblib.load("house_price_model.pkl")

class HouseFeatures(BaseModel):
    bedrooms: int
    bathrooms: float
    sqft_living: int
    sqft_lot: int
    floors: float
    waterfront: int
    view: int
    condition: int
    grade: int
    sqft_above: int
    sqft_basement: int
    yr_built: int
    yr_renovated: int
    zipcode: int
    lat: float
    long: float

@app.post("/predict")
def predict_price(features: HouseFeatures):
    data = np.array([[v for v in features.dict().values()]])
    prediction = model.predict(data)[0]
    return {"predicted_price": round(float(prediction), 2)}

@app.get("/health")
def health():
    return {"status": "ok"}
```
